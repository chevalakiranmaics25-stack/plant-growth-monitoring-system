```python
##Load the cleaned dataset
import pandas as pd

df = pd.read_csv("cleaned_plant_growth.csv")
```

```python
##Select the target column

growth = df["New_Growth_Count"]

print("===== PLANT GROWTH STATISTICAL ANALYSIS =====")
```

```python
##Mean

print("\nMean Growth Count:", growth.mean())
```

```python
##Median

print("Median Growth Count:", growth.median())
```

```python
##Standard Deviation

print("Standard Deviation:", growth.std())
```

```python
##Minimum and Maximum

print("Minimum Growth Count:", growth.min())
print("Maximum Growth Count:", growth.max())
```

```python
##Correlation analysis

print("\n===== CORRELATION WITH NEW GROWTH COUNT =====")

correlation = df.corr(numeric_only=True)

print(
correlation["New_Growth_Count"]
.sort_values(ascending=False)
)
```
